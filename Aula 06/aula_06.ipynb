### Exercício resolvido 1: Filtro de segurança
 
Um data center possui os Servidores $[APP_1, APP_2, APP_3, APP_4]$ e os Bancos $[DB_A, DB_B, DB_C]$. Como poderíamos gerar todas as rotas possíveis entre servidores e bancos de dados? A resposta é: Realizando o produto cartesiano, que pode ser feito utilizando a biblioteca itertools. 

````python
import itertools

servidores = ['APP_1', 'APP_2', 'APP3_3', 'APP_4']
bancos = ['DB_A', 'DB_B', 'DB_C']

print('Servidores: ', servidores)
print('Bancos: ', bancos)

rotas = list(itertools.product(servidores, bancos))
print('\nRotas possíveis: ')
print(rotas)
````

No exercício, existe uma blacklist de rotas definidas por um bloqueio de firewall, que são especificamente: $[(APP_2, DB_B), (APP_4, DB_A)]$, ou seja, de todas as rotas possíveis determinadas pelo produto cartesiano apenas as rotas do Servidor 2 para o Banco B e do Servidor 4 para o Banco A devem ser bloqueadas.

Consideremos $U$ como o conjunto Universo de todas as possibilidades de rotas e o conjunto $B% como sendo o conjunto de rotas da Blacklist, a operação de diferença $U-B$ mostrará todos os conjuntos de pares ordenados que estão no universo de possibilidades $U$ mas que necessariamente não esteja no conjunto de pares que determinam a blacklist. Portanto, aplicando o operador de diferença nos sets de dados acima:

````python
blacklist = [('APP_2', 'DB_B'), ('APP_4', 'DB_A')]
rotas_permitidas = set(rotas)-set(blacklist)
rotas_permitidas_ordenada = sorted(rotas_permitidas, key = rotas.index) #Preserva os índices originais da Lista de rotas
print('Rotas permitidas: ', rotas_permitidas_ordenada)
````

### Exercício resolvido 2: Laudo Automatizado de alta disponibilidade
 
Em arquiteturas de microsserviços, o valor 1 na matriz [linha][coluna] indica que o nó da linha envia dados para o nó da coluna. Para ser certificado, um cluster precisa garantir duas regras:​

* **Self-Health Check**: Todo nó da malha deve possuir uma rota apontando para si mesmo para autodiagnóstico.​
* **Mutual Failover**: Se um nó X envia pacotes para o nó Y, o sistema exige obrigatoriamente que exista uma rota de resposta de Y para X.

Podemos observar que as propriedades elencadas acima são a propriedade Reflexiva, em que cada elemento do conjunto relaciona-se consigo mesmo, e a propriedade Simétrica, que mostra que cada elemento i que relaciona-se com um elemento j deve preservar a relação inversa de j para i, para todos os elementos do conjunto.

Consideremos os seguintes clusters:

```
cluster_alfa = [[0, 1], [1, 1]]​
cluster_beta = [[1, 1], [0, 1]]​
cluster_gama = [[1, 1], [1, 1]]​
```

O exercício pede para criarmos um Script que verifique a integridade dos clusters de acordo com as regras definidas. O Script deve imprimir as seguintes mensagens:

* Se violar o Self-Health Check: 'ALERTA: Falha de autodiagnóstico detectada.'​
* Se violar o Mutual Failover: 'ALERTA: Rota unilateral detectada sem via de retorno.'​
* Se passar em ambos: 'SISTEMA APROVADO.'​

Pede-se para fazer essa verificação para cada cluster e organizar a sua resposta utilizando a função print() para mostrar os resultados.

Pois bem, consideremos os clusters em sua situação atual:

````python
print('Clusters propostos:\n')

cluster_alfa = [
    [0, 1], [1, 1]
]

cluster_beta = [
    [1, 1], [0, 1]
]

cluster_gama = [
    [1, 1], [1, 1]
]

print('Cluster Alfa: ', cluster_alfa)
print('Cluster Beta: ', cluster_beta)
print('Cluster Gama: ', cluster_gama)

#Funções de verificação
def self_health_check(M, n):
    for i in range(n):
        if M[i][i] == 0:
            return False
    return True

def mutual_failover_check(M, n):
    for i in range(n):
        for j in range(i + 1, n):
            if M[i][j] != M[j][i]:
                return False
    return True

print('\nSelf-Health Check:')

````
