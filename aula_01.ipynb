Cenário de aplicação: Consideremos um Sistema de Controle de Acesso baseado em cargos, ou um *RBAC - Role-Based Acess Control -* como sendo a nossa primeira aplicação prática. Consideremos o uso de *sets*, que diferentemente das listas em Python são estruturas desordenadas de itens únicos, ou seja, sem duplicatas. 

Consideremos, primeiro, o uso da estrutura *set* para definir os conjuntos de usuários com especificações bem definidas em nosso exemplo:

```python
desenvolvedores = {'Ana', 'Bruno', 'Carlos', 'Diana'}
analistas_banco = {'Carlos', 'Diana', 'Eduardo'}
administradores = {'Ana', 'Eduardo', 'Fernanda'}
```

Consideremos o seguinte conjunto de regras e especificações que precisamos aplicar:
* Preciso definir quem possui acesso aos servidores de homologação. Os servidores de homologação podem ser acessados por desenvolvedores ou analistas de banco
* Preciso "separar" aqueles colaboradores que são devs e também administradores, para participar de um projeto
* Preciso determinar aqueles que são desenvolvedores, mas não possuem acesso ao banco.

```python

acesso_homologacao = desenvolvedores | analistas_banco
print(f'Acesso à Homologação: {acesso_homologacao}')

dev_admins = desenvolvedores & administradores
print(f'Devs com privilégios de Admin: {dev_admins}')

apenas_devs = desenvolvedores - analistas_banco
print(f'Devs que não possuem acesso ao banco: {apenas_devs}')
```

Neste outro exemplo, vamos utilizar as *bitmasks* para realizar o mapeamento de permissões em um sistema computacional. Consideremos o seguinte mapeamento de permissões:

* 0b001 (1) -> Ler
* 0b010 (2) -> Escrever
* 0b100 (4) -> Executar

Considere o perfil do usuário Carlos, que deve ser de permissçao para ler e executar neste sistema: Como eu posso configurar o sistema para que ele possua essas atribuições?

```python
PERMISSAO_LER = 0b001
PERMISSAO_ESCREVER = 0b010
PERMISSAO_EXECUTAR = 0b100

perfil_carlos = PERMISSAO_LER | PERMISSAO_EXECUTAR # Representação: 101
```

Um exemplo de verificação de permissão específica:

```python
if perfil_carlos & PERMISSAO_LER:
    print('Permissão para leitura.')
else:
    print('Acesso negado para leitura.')

if perfil_carlos & PERMISSAO_ESCREVER:
    print('Permissão para escrita.')
else:
    print('Acesso negado para escrita.')

if perfil_carlos & PERMISSAO_EXECUTAR:
    print('Permissão para executar.')
else:
    print('Permissão para execução.')
```
