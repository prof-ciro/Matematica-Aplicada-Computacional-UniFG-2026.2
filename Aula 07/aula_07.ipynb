Listas são mutáveis e indexadas ordenadamente — crescem e mudam em execução. Tuplas são imutáveis, ideais para constantes que ninguém pode alterar por acidente. Funções encapsulam rotinas: lógica escrita uma vez, reusada sempre. Regra prática: a mutabilidade define onde cada estrutura pode ser usada com segurança.

```python
# Listas
notas = [7.5, 8.0, 9.2]
notas[0] = 6.0     
print("Lista:", notas)

# Tuplas
origem = (0, 0)
# origem[0] = 1         #Retirar o comentário para verificar mensagem de erro
print("Tupla:", origem)

# Funções
def calcular(x):
    return x * 2

print("Função:", calcular(21))
```

### Dicionários

Dicionário é um mapeamento chave → valor: O acesso é pela chave, nunca pelo índice. Um registro pode reunir listas, tuplas e funções. Cada chave do dicionário vai virar uma coluna do DataFrame.

```python
aluno = {
    'nome': 'Ana',
    'notas': [8.5, 7.0, 9.2],   # lista
    'turma': ('CC', 2025)       # tupla
}

def media(a):                   # função
    return sum(a['notas']) / len(a['notas'])

print(round(media(aluno), 1))   # 8.2
```

### Revisão de Programação Orientada a Objetos - POO

Vamos revisar a nomenclatura que utilizaremos. É importante estar bastante familiarizado:

* Uma Classe é o molde;
* Um Objeto é a instância;
* Atributos guardam dados;
* Métodos guardam comportamentos.

Por exemplo: df.head() é um método do objeto DataFrame — a mesma ideia da classe Aluno.

```python
class Aluno:                        
    def __init__(self, nome, notas):
        self.nome = nome
        self.notas = notas

    def media(self):                
        n = self.notas
        return sum(n) / len(n)

ana = Aluno('Ana', [8.5, 7.0, 9.2])
print(round(ana.media(), 1))        # 8.23
```

### Series e DataFrames

Series é um vetor unidimensional (uma coluna com índice), enquanto que o DataFrame é a estrutura matricial completa. Na prática, um DataFrame é um dicionário em que cada chave é uma coluna e cada valor é uma Series.

```python
import pandas as pd

dados = {
    'nome':  ['Ana', 'Bruno', 'Carla', 'Diego'],
    'idade': [22, 25, 19, 23],
    'turma': ['CC', 'ADS', 'CC', 'ADS'],
    'nota':  [8.5, 7.0, 9.2, 6.8]
}

df = pd.DataFrame(dados)
print(df)
print("Tipo da coluna:", type(df['nome']))
print("Tipo da tabela:", type(df))
```

### Máscaras Booleanas

A avaliação lógica retorna um vetor de True/False. Aplicada ao DataFrame, a máscara devolve só as linhas verdadeiras — substituindo o if escrito à mão. A condição matemática é aplicada à coluna inteira de uma vez.

```python

mascara = df['idade'] > 20
print(mascara)

print('\n\n')

print(df[mascara])

print('\n\n')

print(df[df['idade'] > 20])

print('\n\n')
```
