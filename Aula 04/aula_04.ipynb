## Cenário de aplicação: Implementação de algoritmos de auditoria matricial no Python aplicando as regras de identificação que acabaram de ser deduzidas no quadro.

### Explicação Estrutural do Algoritmo

Reflexiva: 
* $M[i][i] = 1$ para todo $i$.
* Simétrica: $M[i][j] = M[j][i]$.
* Anti-simétrica: Se $M[i][j] = 1$ e $i \neq j$, então obrigatoriamente $M[j][i] = 0$
* Transitiva: Se $M[i][j]=1$ e $M[j][k]=1$, então deve existir $M[i][k]=1$.

```python
# ==========================================
# 1. REVISÃO: GERAÇÃO DA MATRIZ BOOLEANA
# ==========================================

# Domínio S
S = [0, 1, 2, 3]
n = len(S)

# Definição da relação exemplo
relacao = [
    (0, 0), (0, 1), (0, 3),
    (1, 0), (1, 1),
    (2, 2),
    (3, 0), (3, 3)
]

# Inicializando a matriz quadrada n x n com zeros
matriz = [[0 for _ in range(n)] for _ in range(n)]

# Consigurando matriz Booleana para reproduzir a relação exemplo
for i, x in enumerate(S):
    for j, y in enumerate(S):
        if (x, y) in relacao:
            matriz[i][j] = 1

print("--- Matriz Booleana Gerada ---")
for linha in matriz:
    print(linha)
print("\n")

# ==========================================
# 2. ALGORITMOS DE AUDITORIA E VERIFICAÇÃO
# ==========================================
def verifica_reflexiva(M, n):
    for i in range(n):
        if M[i][i] == 0:
            return False
    return True

def verifica_simetrica(M, n):
    for i in range(n):
        for j in range(n):
            if M[i][j] != M[j][i]:
                return False
    return True

def verifica_anti_simetrica(M, n):
    for i in range(n):
        for j in range(n):
            if i != j and M[i][j] == 1 and M[j][i] == 1:
                return False
    return True

def verifica_transitiva(M, n):
    for i in range(n):
        for j in range(n):
            if M[i][j] == 1:
                for k in range(n):
                    if M[j][k] == 1 and M[i][k] == 0:
                        return False
    return True

# ==========================================
# 3. EXECUÇÃO DA IDENTIFICAÇÃO ESTRUTURAL
# ==========================================
print("--- Relatório de Auditoria e Verificação ---")
print(f"Reflexiva: {verifica_reflexiva(matriz, n)}")
print(f"Simétrica: {verifica_simetrica(matriz, n)}")
print(f"Anti-simétrica: {verifica_anti_simetrica(matriz, n)}")
print(f"Transitiva: {verifica_transitiva(matriz, n)}")
```
