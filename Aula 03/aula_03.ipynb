Cenário de aplicação: Consideremos um Sistema Acadêmico que fará o registro dos alunos, que podem cursar dois cursos: Python ou Hardware. 

### Explicação Estrutural do Algoritmo

**1. Definição de Domínios e Construção do Produto Cartesiano**
O algoritmo inicia com a importação do módulo `itertools`, que fornece funções eficientes para iteração. Os domínios do problema são declarados em estruturas estáticas de dados: o conjunto $A$ (alunos) e o conjunto $B$ (cursos). Através da função `itertools.product`, computa-se o produto cartesiano $A \times B$, gerando exaustivamente todos os pares ordenados (tuplas) possíveis entre os dois conjuntos, o que compõe o espaço amostral completo das combinações no sistema.

**2. Delimitação da Relação Binária ($\rho$)**
A variável `relacao_matricula` materializa a relação binária. Trata-se de um subconjunto estrito do produto cartesiano previamente gerado, cuja função é filtrar e reter unicamente as tuplas que satisfazem a regra de negócio preestabelecida — neste caso, os discentes efetivamente associados às disciplinas correspondentes.

**3. Alocação da Matriz Booleana via *List Comprehension***
Para a representação computacional desta relação, utiliza-se a técnica de *list comprehension* na inicialização de uma matriz bidimensional. O processamento aloca uma matriz de ordem $m \times n$, onde $m$ corresponde à cardinalidade do conjunto $A$ e $n$ à cardinalidade do conjunto $B$. Inicialmente, todos os índices recebem a atribuição do escalar `0`, definindo o estado lógico padronizado de ausência de relação para todas as coordenadas matriciais.

**4. Mapeamento Condicional e Atribuição Lógica**
Uma estrutura de repetição aninhada itera sobre os domínios. A função `enumerate()` atua na captura direta do elemento nominal e de seu respectivo índice posicional ($i$ para a dimensão das linhas, $j$ para a dimensão das colunas). A avaliação condicional `if` confere a pertinência lógica da tupla corrente `(aluno, curso)` à relação definida no subconjunto matricial. Quando a condição de pertinência retorna valor lógico verdadeiro (True), a coordenada exata $[i][j]$ recebe o nível lógico `1`, registrando matematicamente a existência da relação no modelo computacional.

**5. Validação da Estrutura Resultante**
O bloco final percorre a matriz para a transcrição formatada dos dados alocados em memória. O arranjo estrutural viabiliza que o sistema abstraia consultas de busca linear iterativa em favor de verificações pontuais de índice bidimensional, otimizando o acesso e a manipulação dos dados relacionais com complexidade temporal constante $O(1)$.

O código exemplo segue abaixo:

```python
import itertools

# 1. ELEMENTOS QUE COMPORÃO A RELAÇÃO, OU SEJA, OS CONJUNTOS A E B
alunos = ['Ana', 'Bruno', 'Carlos']
cursos = ['Python', 'Hardware']

# 2. O PRODUTO CARTESIANO (A x B) - COMBINAÇÃO INTEGRAL ENTRE OS ELEMTNOS DOS CONJUNTOS A E B
dominio_total = list(itertools.product(alunos, cursos))
print('Produto cartesiano: ')
print(dominio_total)

# 3. A RELAÇÃO (DESCRIÇÃO DE QUEM SE RELACIONA COM QUEM)
relacao_matricula = [
    ('Ana', 'Python'), 
    ('Ana', 'Hardware'), 
    ('Carlos', 'Hardware')
]

# 4. TRADUÇÃO PARA MATRIZ BOOLEANA
matriz = [[0 for _ in cursos] for _ in alunos] #CRIAÇÃO DE ESTRUTURA

for i, aluno in enumerate(alunos):
    for j, curso in enumerate(cursos):
        if (aluno, curso) in relacao_matricula:
            matriz[i][j] = 1

print('\nMatriz Booleana por aluno:')
for i, linha in enumerate(matriz):
    print(f'{alunos[i]}: {linha}')
```

Percebemos que as matrizes booleanas indicam a *flag* 1 para o caso em que o aluno está matriculado no curso, seguindo a estrutura `[flag_python flah_hardware]`. Percebemos a matriz `[1,1]` indicando que a Ana está matriculada em ambos os cursos, ao passo que o Bruno não está matriculado em nenhunm, e o Carlos apenas em Hardware.
